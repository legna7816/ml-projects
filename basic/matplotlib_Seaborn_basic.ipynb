```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    '이름': ['철수', '영희', '민수', '지은', '현우'],
    '나이': [23, 25, np.nan, 22, 24],
    '학과': ['컴공', '경영', '컴공', np.nan, '전자'],
    '학점': [3.5, 4.1, 3.8, 2.9, np.nan]
}
df = pd.DataFrame(data)
df['학점_보정'] = df['학점'].fillna(0)

plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(6, 4))
plt.bar(df['이름'], df['학점_보정'])
plt.xlabel('이름')
plt.ylabel('학점')
plt.title('이름별 학점')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df['나이'].dropna(), bins=5)
plt.xlabel('나이')
plt.ylabel('빈도')
plt.title('나이 분포')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(df['나이'], df['학점_보정'])
plt.xlabel('나이')
plt.ylabel('빈도')
plt.title('나이 분포')
plt.show()
```

```python
dept_mean = df.groupby('학과')['학점_보정'].mean()

plt.figure(figsize=(6, 4))
dept_mean.plot(kind='bar')
plt.ylabel('평균 학점')
plt.title('학과별 평균 학점')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='학과', y='학점_보정')
plt.title('학과별 평균 학점 (seaborn)')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x='학과', y='학점_보정')
plt.title('학과별 학점 분포 (박스플롯)')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='학과')
plt.title('학과별 인원수')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x='나이', y='학점_보정')
plt.title('나이별 학점 분포')
plt.show()
```

```python
dept_mean=df.groupby('학과')['나이'].mean()

plt.figure(figsize=(6, 4))
dept_mean.plot(kind='bar')
plt.ylabel('나이 평균')
plt.title('학과별 나이 평균')
plt.show()
```
