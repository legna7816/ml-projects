```python
import pandas as pd
import numpy as np
```

```python
data = {
    '이름': ['철수', '영희', '민수', '지은', '현우'],
    '나이': [23, 25, np.nan, 22, 24],
    '학과': ['컴공', '경영', '컴공', np.nan, '전자'],
    '학점': [3.5, 4.1, 3.8, 2.9, np.nan]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

```python
print(df['이름'])
print(df[['이름', '학점']])
```

```python
print(df.loc[0])
print(df.iloc[0])
```

```python
print(df[df['학점'] >= 3.5])
print(df[(df['학점'] >= 3.5) & (df['학과'] == '컴공')])
```

```python
print(df.isnull())
print(df.isnull().sum())
```

```python
df_dropped = df.dropna()
df_filled_age = df['나이'].fillna(df['나이'].mean())
df_filled_dept = df['학과'].fillna('미정')

print(df_dropped)
print(df_filled_age)
```

```python
print(df.groupby('학과')['학점'].mean())
print(df.groupby('학과').agg({'학점': 'mean', '나이': 'mean'}))
print(df['학과'].value_counts())
```

```python
print(df[df['나이']>=23])
print(df.groupby('학과')['나이'].max())
df['학점_보정'] = df['학점'].fillna(0)
print(df['학점_보정'])
```
