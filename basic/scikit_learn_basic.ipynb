```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris, make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, mean_squared_error, r2_score
```

```python
iris = load_iris()
X = iris.data
y = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)
df['species'] = y
print(df.head())
print(df['species'].value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)
print(X_train.shape, X_test.shape)
```

```python
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(y_pred[:10])
print(y_test[:10])
```

```python
print('정확도: ', accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
tree_model = DecisionTreeClassifier(max_depth=3)
tree_model.fit(X_train, y_train)
tree_pred = tree_model.predict(X_test)

print('decisionTree 정확도: ', accuracy_score(y_test, tree_pred))
```

```python
X_reg, y_reg = make_regression(n_samples=200, n_features=1, noise=15, random_state=42)
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg, y_reg, test_size=0.3, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_reg_train, y_reg_train)
y_reg_pred = reg_model.predict(X_reg_test)

print('MSE: ', mean_squared_error(y_reg_test, y_reg_pred))
print('R2: ', r2_score(y_reg_test, y_reg_pred))
```

```python
from sklearn.model_selection import cross_val_score

log_scores = cross_val_score(LogisticRegression(max_iter=200), X, y, cv=5)
tree_scores = cross_val_score(DecisionTreeClassifier(random_state=42), X, y, cv=5)

print('Logistic 평균:', log_scores.mean(), log_scores)
print('Tree 평균:', tree_scores.mean(), tree_scores)
```
