```python
# 1. 사전학습된 임베딩 불러오기 (gensim)
import gensim.downloader as api

print("임베딩 다운로드 중")
wv = api.load('glove-wiki-gigaword-100')
print('완료! 단어 개수:', len(wv))
```

```python
# 2. 단어를 벡터로 확인
print("'king'의 벡터 (앞 10개 값):")
print(wv['king'][:10])
print("벡터 차원:", wv['king'].shape)
```

```python
# 3. 의미적 유사도 - TF-IDF로는 절대 불가능했던 것
# 비슷한 뜻의 단어 찾기
print("\n'good'과 비슷한 단어:")
print(wv.most_similar('good', topn=5))

print("\n'movie'와 비슷한 단어")
print(wv.most_similar('movie', topn=5))

# 두 단어 사이 유사도 (0~1, 1에 가까울수록 비슷)
print("\ngood vs great", wv.similarity('good', 'great'))
print("good vs terrible", wv.similarity('good', 'terrible'))
```

```python
# 4. 유명한 의미 연산: king - man + woman = ?
result = wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=3)
print("\nking - man + woman = ", result)
```
